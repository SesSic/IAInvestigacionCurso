### 1.1 ListasUna lista guarda varios valores en orden. Ejemplo con **mediciones de ingeniería** (resistencia de 5 probetas, en MPa):

```python
resistencias = [24.5, 26.1, 23.8, 25.0, 24.9]print(resistencias)print("Primer valor:", resistencias[0])print("Cantidad de datos:", len(resistencias))
```

### 1.2 DiccionariosUn diccionario guarda pares `clave: valor` — ideal para representar un registro (ej. un paciente, un caso, un estudiante). Ejemplo de **salud** (signos vitales de un paciente):

```python
paciente = {    "nombre": "Paciente 1",    "presion_sistolica": 120,    "frecuencia_cardiaca": 78,    "temperatura": 36.7}print("Frecuencia cardíaca:", paciente["frecuencia_cardiaca"])
```

### 1.3 FuncionesUna función guarda un bloque de código para reutilizarlo. Ejemplo de **educación**: función que decide si un estudiante aprueba.

```python
def esta_aprobado(nota):    return nota >= 7# Prueba tu funciónprint(esta_aprobado(8.5))print(esta_aprobado(5.0))
```

---## 🟩 Bloque 2 (50 min) — NumPy: operar sobre muchos datos a la vez

### 2.1 ¿Por qué NumPy?Con listas normales, para sumar dos conjuntos de datos necesitas un bucle. Con NumPy, lo haces directo, como si fueran columnas de Excel.

```python
import numpy as npnotas_parcial1 = np.array([7.5, 6.8, 8.9, 5.5, 9.0])notas_parcial2 = np.array([8.2, 7.0, 9.1, 6.2, 9.4])promedio_final = (notas_parcial1 + notas_parcial2) / 2print(promedio_final)
```

### 2.2 Funciones útiles de NumPy

```python
print("Promedio general:", np.mean(promedio_final))print("Nota más alta:", np.max(promedio_final))print("Nota más baja:", np.min(promedio_final))print("Desviación estándar:", np.std(promedio_final))
```

### 3.1 Leer el archivo de la Sesión 1Vamos a retomar `ejemplo_notas.csv` (o tu propio archivo) y esta vez usarlo para automatizar algo que antes hacías a mano.

```python
import pandas as pdurl = "https://raw.githubusercontent.com/TU-USUARIO/curso-programacion-cientifica-ia/main/datasets/ejemplo_notas.csv"datos = pd.read_csv(url)datos.head()
```

### 3.2 Automatizar con un bucleImagina que necesitas revisar, uno por uno, si cada estudiante aprobó — algo que en Excel harías con una fórmula copiada 60 veces. Aquí lo haces con un bucle que revisa todos a la vez:

```python
for i, fila in datos.iterrows():    nombre = fila["estudiante"]    nota = fila["nota_parcial1"]    resultado = esta_aprobado(nota)    print(nombre, "->", "Aprobado" if resultado else "Reprobado")
```

### 3.3 La forma "profesional" de hacerlo (sin bucle explícito)Pandas te permite aplicar una función a toda una columna de una vez, más rápido y con menos código:

```python
datos["aprobado"] = datos["nota_parcial1"].apply(esta_aprobado)datos[["estudiante", "nota_parcial1", "aprobado"]]
```

### 1.1 Listas
Una lista guarda varios valores en orden.

Ejemplo con **mediciones de ingeniería** (resistencia de 5 probetas, en MPa):

```python
resistencias = [24.5, 26.1, 23.8, 25.0, 24.9]
print(resistencias)
print("Primer valor:", resistencias[0])
print("Cantidad de datos:", len(resistencias))
```

### 1.2 Diccionarios
Un diccionario guarda pares `clave: valor` — ideal para representar un registro (ej. un paciente, un caso, un estudiante).

Ejemplo de **salud** (signos vitales de un paciente):

```python
paciente = {"nombre": "Paciente 1",    "presion_sistolica": 120,    "frecuencia_cardiaca": 78,    "temperatura": 36.7}
print("Frecuencia cardíaca:", paciente["frecuencia_cardiaca"])
```

### 1.3 Funciones
Una función guarda un bloque de código para reutilizarlo.

Ejemplo de **educación**: función que decide si un estudiante aprueba.

```python
def esta_aprobado(nota):
  return nota >= 7
  # Prueba tu función
  print(esta_aprobado(8.5))
  print(esta_aprobado(5.0))
```

---## 🟩 Bloque 2 (50 min) — NumPy: operar sobre muchos datos a la vez

### 2.1 ¿Por qué NumPy?
Con listas normales, para sumar dos conjuntos de datos necesitas un bucle. Con NumPy, lo haces directo, como si fueran columnas de Excel.

```python
import numpy as np
notas_parcial1 = np.array([7.5, 6.8, 8.9, 5.5, 9.0])
notas_parcial2 = np.array([8.2, 7.0, 9.1, 6.2, 9.4])
promedio_final = (notas_parcial1 + notas_parcial2) / 2
print(promedio_final)
```

### 2.2 Funciones útiles de NumPy

```python
print("Promedio general:", np.mean(promedio_final))
print("Nota más alta:", np.max(promedio_final))
print("Nota más baja:", np.min(promedio_final))
print("Desviación estándar:", np.std(promedio_final))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
url = "https://raw.githubusercontent.com/SesSic/IAInvestigacionCurso/main/datasets/ejemplo_notas.csv"
datos = pd.read_csv(url)
datos.head()
```

---## 🟦 Bloque 1 (50 min) — Preparar los datos

### 1.1 La idea central del aprendizaje automático

Un modelo aprende un patrón a partir de datos que **ya conoces la respuesta** (datos de entrenamiento), y luego lo usas para predecir sobre datos nuevos donde no la conoces.
- **Variable objetivo (`y`):** lo que quieres predecir
- **Variables predictoras (`X`):** la información que usas para predecir

### 1.2 Separar variables predictoras y objetivo

Ejemplo: predecir `nota_parcial2` (objetivo) usando `nota_parcial1` y `asistencia_pct` (predictoras).

```python
X = datos[["nota_parcial1", "asistencia_pct"]]
y = datos["nota_parcial2"]
print(X.head())
print(y.head())
```

### 1.3 Dividir en entrenamiento y prueba

Nunca evalúas un modelo con los mismos datos que usaste para entrenarlo — sería como calificar un examen con las respuestas ya dadas.

Se reserva una parte (típicamente 20-30%) exclusivamente para probar.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("Datos de entrenamiento:", len(X_train))
print("Datos de prueba:", len(X_test))
```

---## 🟩 Bloque 2 (50 min) — Regresión y clasificación

### 2.1 Regresión: predecir un número

Vamos a entrenar un modelo simple que prediga `nota_parcial2` a partir de `nota_parcial1` y `asistencia_pct`.

```python
from sklearn.linear_model import LinearRegression
modelo_regresion = LinearRegression()
modelo_regresion.fit(X_train, y_train)
predicciones = modelo_regresion.predict(X_test)
print("Predicciones:", predicciones[:5])
print("Valores reales:", y_test.values[:5])
```

### 2.2 Clasificación: predecir una categoría

Ahora vamos a predecir si alguien **aprueba o no** (categoría), usando un árbol de decisión — un modelo fácil de interpretar porque puedes "leer" sus reglas de decisión.

```python
from sklearn.tree import DecisionTreeClassifier
datos["aprobado"] = (datos["nota_parcial1"] >= 7).astype(int)
X_clas = datos[["asistencia_pct"]]
y_clas = datos["aprobado"]
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_clas, y_clas, test_size=0.3, random_state=42)
modelo_clasificacion = DecisionTreeClassifier(max_depth=3, random_state=42)
modelo_clasificacion.fit(X_train_c, y_train_c)
predicciones_clas = modelo_clasificacion.predict(X_test_c)
print("Predicciones:", predicciones_clas)
print("Valores reales:", y_test_c.values)
```

---## 🟨 Bloque 3 (50 min) — Evaluar qué tan bueno es el modelo

### 3.1 Métricas para regresión
- **Error absoluto medio (MAE):** en promedio, ¿cuánto se equivoca el modelo? (en las mismas unidades que tu variable)
- **R²:** ¿qué tan bien explica el modelo la variación de los datos? (cercano a 1 es mejor, cercano a 0 es malo)

```python
from sklearn.metrics import mean_absolute_error, r2_score
mae = mean_absolute_error(y_test, predicciones)
r2 = r2_score(y_test, predicciones)
print("Error absoluto medio:", round(mae, 2))
print("R²:", round(r2, 2))
```

### 3.2 Métricas para clasificación
- **Exactitud (accuracy):** ¿qué porcentaje de predicciones acertó el modelo?

```python
from sklearn.metrics import accuracy_score
exactitud = accuracy_score(y_test_c, predicciones_clas)
print("Exactitud del modelo:", round(exactitud * 100, 1), "%")
```

### 3.3 ⚠️ Cuidado con datasets pequeños

Con pocos datos (como nuestro ejemplo de 15 filas), estas métricas no son muy confiables — un modelo real de investigación necesita muchos más datos para que las métricas signifiquen algo sólido. Hoy el objetivo es que entiendas el **proceso completo** (preparar → entrenar → predecir → evaluar), que es igual sin importar el tamaño de los datos.

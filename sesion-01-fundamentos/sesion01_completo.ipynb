---### 🧪 Ejercicio 2 — Estadísticas básicas de tu tabla

Usa las columnas numéricas de tu tabla (o de `ejemplo_notas.csv`) para responder:

```python
# ¿Cuál es el promedio de la columna 'nota_parcial1'?
promedio_p1 = datos["nota_parcial1"].mean()
print("Promedio parcial 1:", promedio_p1)

# ¿Cuál es el valor máximo de 'nota_parcial2'?
maximo_p2 = datos["nota_parcial2"].max()
print("Máximo parcial 2:", maximo_p2)
```

---## 🟨 Bloque 3 (50 min) — Primer análisis por grupo + cierre

### 3.1 Agrupar y comparar

Una de las tareas más comunes en investigación: comparar promedios entre grupos (por ejemplo, grupo A vs grupo B, o facultad vs facultad).

```python
promedio_por_grupo = datos.groupby("grupo")["nota_parcial1"].mean()
print(promedio_por_grupo)
```

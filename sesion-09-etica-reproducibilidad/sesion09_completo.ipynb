### 1.1 ¿Qué es un sesgo en datos?

Un modelo o análisis "aprende" lo que hay en los datos — si los datos reflejan una desigualdad o subrepresentación, el resultado la reproduce y a veces la amplifica.

**Ejemplos reales:**
- Un modelo de diagnóstico médico entrenado mayormente con datos de un grupo etario o género puede fallar en otros grupos
- Un sistema de selección de personal entrenado con contrataciones históricas puede replicar sesgos de género o edad de esas contrataciones
- Una base de "casos exitosos" en educación que solo incluye una facultad no representa a las demás

### 1.3 Privacidad: qué NO subir a herramientas de IA públicas

Regla simple: **si no lo publicarías en una red social, no lo pegues en ChatGPT/Claude/DeepSeek/Gemini gratuitos.**

**Nunca subas a estas herramientas:**
- Nombres reales de pacientes, historias clínicas, diagnósticos identificables
- Datos de menores de edad identificables
- Información de casos legales en curso con datos de las partes
- Cualquier dato que tu institución clasifique como confidencial

**Sí puedes usar (anonimizado):**
- Datos con nombres reemplazados por códigos ("Paciente 001" en vez del nombre real)
- Estadísticas agregadas (promedios, porcentajes) sin identificar personas individuales

---## 🟩 Bloque 2 (50 min) — Autoría y transparencia

### 2.1 ¿Puedes poner a la IA como autora de un paper?
**No.**

Las políticas de las principales editoriales (Nature, IEEE, Elsevier, entre otras) coinciden en que la IA **no puede ser autora** porque no puede asumir responsabilidad por el contenido. Pero sí exigen que **declares su uso** si la usaste para redactar, analizar o generar código.

### 2.2 Cómo declarar el uso de IA (ejemplo de texto)

> "Durante la preparación de este trabajo, el/los autor(es) usó/usaron [nombre de la herramienta] para [tarea específica: redacción, traducción, generación de código, análisis exploratorio]. Después de usar esta herramienta, el/los autor(es) revisó/revisaron y editó/editaron el contenido según fue necesario, y asume/asumen total responsabilidad por el contenido de la publicación."

Esto normalmente va en la sección de **Agradecimientos** o **Métodos**, según la revista.

---## 🟨 Bloque 3 (50 min) — Git/GitHub: crea tu propio repositorio

### 3.1 ¿Por qué esto es parte de "ética"?

La reproducibilidad **es** un compromiso ético: cualquier persona debería poder revisar tu trabajo y llegar al mismo resultado. Un análisis que solo existe en tu computador, sin documentar, no es verificable por nadie más — ni siquiera por ti mismo dentro de 6 meses.

### 3.2 Crea tu cuenta y repositorio (si no lo tienes ya)
1. Ve a **github.com** → crea una cuenta gratuita si no tienes
2. Clic en **"New"** → nombre sugerido: `mi-investigacion-[tu-tema]`
3. Marca **Public**
4. Clic en **Create repository**

### 3.3 Sube tu trabajo del curso

Sube al menos:

### 3.4 Ejemplo de un repositorio de investigación real

Un proyecto de investigación bien documentado normalmente organiza carpetas así:

```
proyecto-investigacion/
├── README.md              ← qué es, cómo reproducirlo
├── datos/                 ← datos crudos y procesados
├── notebooks/             ← análisis paso a paso
├── resultados/            ← gráficos, tablas finales
└── LICENSE                ← términos de uso del código
```

Esta misma lógica aplica sea tu proyecto de robótica, un estudio clínico, o una revisión de sentencias judiciales — la organización clara es lo que hace que el trabajo sea reutilizable por ti o por otros.

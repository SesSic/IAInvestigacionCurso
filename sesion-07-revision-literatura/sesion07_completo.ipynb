---## 🟨 Bloque 3 (50 min) — Verificación de referencias y cierre

### 3.1 ⚠️ El riesgo más importante de esta sesión: alucinaciones de referencias

Las IA generativas a veces **inventan papers que no existen** — con título, autores y año que suenan completamente reales, pero que no puedes encontrar en ningún lado.

Esto se llama "alucinación", y es especialmente común cuando le pides a la IA que **te recomiende** bibliografía (en vez de que analice un texto que tú le diste).

### 3.2 Cómo verificar una referencia

Antes de citar cualquier referencia que te dé una IA, verifica:
1. **Busca el título exacto** en Google Scholar o Scopus — ¿aparece?
2. **Verifica el DOI** (si te lo dio) en doi.org — ¿lleva a un artículo real?
3. **Confirma que los autores existan** y hayan publicado sobre ese tema
